````python
from google import genai

def callGPT(prompt):
  client = genai.Client(api_key="APIKEY")
  response = client.models.generate_content(
      model="gemini-2.5-flash",
      contents=prompt
  )

  return response
````

````python
def cleanse_response(response):
  clean_response=clean_response = response.text
  clean_response = clean_response.replace("```json", "")
  clean_response = clean_response.replace("```", "")
  clean_response = clean_response.strip()

  return clean_response
````

````python
tools = [
    "find_orphaned_accounts",
    "find_dormant_accounts",
    "calculate_jml_metrics",
    "get_source_health"
]
````

````python
question = """
Generate an IAM health report
"""

planner_prompt = f"""
You are an IAM Planner.

Available tools:

{tools}

Question:

{question}

Return JSON only.

Example:

{{
  "plan": [
    "find_orphaned_accounts",
    "find_dormant_accounts"
  ]
}}
"""
````

````python
response_json=cleanse_response(callGPT(planner_prompt))
````

````python
import json



responseObj = json.loads(response_json)

plan  = responseObj["plan"]

print(plan)

````

````python
def find_orphaned_accounts():
    return 12

def find_dormant_accounts():
    return 45

def calculate_jml_metrics():
    return {"avg_hours": 8}

def get_source_health():
    return "GREEN"
````

````python
tool_map = {
    "find_orphaned_accounts": find_orphaned_accounts,
    "find_dormant_accounts": find_dormant_accounts,
    "calculate_jml_metrics": calculate_jml_metrics,
    "get_source_health": get_source_health
}

results = {}

for step in plan:
    results[step] = tool_map[step]()
````

````python
print(results)
````
