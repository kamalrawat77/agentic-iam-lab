````python
from google import genai

def callGPT(prompt):
  client = genai.Client(api_key="APIKEY")
  response = client.models.generate_content(
      model="gemini-2.5-flash",
      contents=prompt
  )

  return response
````

````python
def cleanse_response(response):
  clean_response=clean_response = response.text
  clean_response = clean_response.replace("```json", "")
  clean_response = clean_response.replace("```", "")
  clean_response = clean_response.strip()

  return clean_response
````

````python
evidence = """
Dormant Accounts: 95

Privileged Dormant Accounts: 30

JML Average Time: 36 Hours

Source Health: GREEN
"""
````

````python
reasoner_prompt = f"""
You are an IAM analyst.

Evidence:

{evidence}

Provide:

1. Findings
2. Risks
3. Recommendations
"""
````

````python
analysis = callGPT(reasoner_prompt).text
````

````python
print(analysis)
````

````python
critic_prompt = f"""
You are a senior IAM auditor.

Evidence:

{evidence}

Analysis:

{analysis}

Review the analysis.

Answer:

1. Is the analysis supported by evidence?
2. What claims lack evidence?
3. What additional information would improve confidence?

Be critical.
"""
````

````python
critic_response = callGPT(critic_prompt).text
````
