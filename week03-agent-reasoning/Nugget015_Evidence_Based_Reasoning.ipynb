```python
from google import genai

def callGPT(prompt):
  client = genai.Client(api_key="APIKEY")
  response = client.models.generate_content(
      model="gemini-2.5-flash",
      contents=prompt
  )

  return response
```

```python
iam_report = {
    "orphaned_accounts": 42,
    "dormant_accounts": 95,
    "jml_avg_hours": 32,
    "source_health": "GREEN"
}
```

```python
evidence = [
    f"Orphaned Accounts: {iam_report['orphaned_accounts']}",
    f"Dormant Accounts: {iam_report['dormant_accounts']}",
    f"Average JML Time: {iam_report['jml_avg_hours']} hours",
    f"Source Health: {iam_report['source_health']}"
]

print(evidence)
```

```python
prompt = f"""
You are a senior IAM analyst.

Evidence:

{evidence}

Instructions:

1. List observations.
2. Explain why they matter.
3. Provide a risk assessment.

Every conclusion must be supported by evidence.
"""
```

```python
print(callGPT(prompt).text)
```

```python
prompt = f"""
You are a senior IAM analyst.

Evidence:

{evidence}

Rules:

- Never invent facts.
- Every conclusion must cite evidence.
- If evidence is insufficient, say so.

Output format:

Evidence:
...

Conclusion:
...

Recommendation:
...
"""
```

```python
print(callGPT(prompt).text)
```
