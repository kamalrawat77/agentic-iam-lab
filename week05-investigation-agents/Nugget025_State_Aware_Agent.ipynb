````python
from google import genai

def callGPT(prompt):
  client = genai.Client(api_key="APIKEY")
  response = client.models.generate_content(
      model="gemini-2.5-flash",
      contents=prompt
  )

  return response
````

````python
def cleanse_response(response):
  clean_response=clean_response = response.text
  clean_response = clean_response.replace("```json", "")
  clean_response = clean_response.replace("```", "")
  clean_response = clean_response.strip()

  return clean_response
````

````python
import json

def return_json(clean_response,objname):
  responseObj = json.loads(clean_response)
  resJsonObj  = responseObj[objname]
  return resJsonObj

````

Create previous investigation and current investigation

````python
previous_investigation = {
    "dormant_accounts": 50,
    "inactive_approvers": 20,
    "sla_compliance": 85
}
current_investigation = {
    "dormant_accounts": 70,
    "inactive_approvers": 35,
    "sla_compliance": 72
}
````

````python
def calculate_change(current, previous):

    return current - previous
````

````python
change = calculate_change(
    current_investigation["dormant_accounts"],
    previous_investigation["dormant_accounts"]
)

print(change)
````

````python
trend_summary = {
    "dormant_change":
        current_investigation["dormant_accounts"]
        - previous_investigation["dormant_accounts"],

    "approver_change":
        current_investigation["inactive_approvers"]
        - previous_investigation["inactive_approvers"],

    "sla_change":
        current_investigation["sla_compliance"]
        - previous_investigation["sla_compliance"]
}

print(trend_summary)
````

````python
prompt = f"""
You are an IAM investigation agent.

Previous Investigation:

{previous_investigation}

Current Investigation:

{current_investigation}

Trend Summary:

{trend_summary}

Generate:

1. Key Changes
2. Emerging Risks
3. Recommended Actions

Focus on trends, not snapshots.
"""
````

````python
print(callGPT(prompt).text)
````
