````python
from google import genai

def callGPT(prompt):
  client = genai.Client(api_key="APIKEY")
  response = client.models.generate_content(
      model="gemini-2.5-flash",
      contents=prompt
  )

  return response
````

````python
def cleanse_response(response):
  clean_response=clean_response = response.text
  clean_response = clean_response.replace("```json", "")
  clean_response = clean_response.replace("```", "")
  clean_response = clean_response.strip()

  return clean_response
````

````python
evidence = {
    "dormant_accounts": 95,
    "privileged_dormant": 30,
    "trend_data": True,
    "source_health": "GREEN"
}
````

````python
def evidence_quality(evidence):

    score = 0

    if "dormant_accounts" in evidence:
        score += 1

    if "privileged_dormant" in evidence:
        score += 1

    if evidence.get("trend_data"):
        score += 1

    if evidence.get("source_health"):
        score += 1

    return score
````

````python
score = evidence_quality(evidence)

print(score)
````

````python
def quality_label(score):

    if score >= 4:
        return "HIGH"

    elif score >= 2:
        return "MEDIUM"

    return "LOW"
````

````python
print(quality_label(score))
````

````python
prompt = f"""
You are a senior IAM investigator.

Evidence:

{evidence}

Evidence Quality:

{quality_label(score)}

Explain:

1. What evidence exists
2. What evidence is missing
3. How missing evidence affects confidence
"""
````

````python
print(callGPT(prompt).text)
````
