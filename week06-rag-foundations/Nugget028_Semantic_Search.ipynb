````python
from google import genai

def callGPT(prompt):
  client = genai.Client(api_key="APIKEY")
  response = client.models.generate_content(
      model="gemini-2.5-flash",
      contents=prompt
  )

  return response
````

````python
semantic_map = {
    "approval delays":"Inactive Approvers",
    "manager not responding":"Inactive Approvers",
    "workflow issue":"Workflow Bottleneck",
    "slow provisioning":"Provisioning Delays"
}
````

````python
def semantic_search(query):

    mapped = semantic_map.get(
        query.lower()
    )

    return mapped
````

````python
print(
    semantic_search(
        "approval delays"
    )
)
````
