````python
with open("investigation_memory.json") as f:
    investigations = json.load(f)

print(investigations)
````

Create searchable text.

````python
documents = []

for inv in investigations:

    documents.append(
        f"""
        Root Cause:
        {inv['root_cause']}

        Recommendation:
        {inv['recommendation']}
        """
    )
````

Generate Embeddings

````python
embeddings = model.encode(documents)
````

Create retrieval function and test it

````python
def retrieve(query):

    query_embedding = model.encode(query)

    scores = util.cos_sim(
        query_embedding,
        embeddings
    )

    best_index = scores.argmax()

    return documents[best_index]
````

````python
def retrieve_top5(query):

    query_embedding = model.encode(query)

    scores = util.cos_sim(
        query_embedding,
        embeddings
    )

    top_indices = scores[0].argsort(descending=True)[:5]

    contexts = []

    for idx in top_indices:
        contexts.append(documents[idx])

    #context = "\n\n".join(contexts)
    return contexts
````

````python
query = "Why are manager approvals delayed?"
candidate_docs=retrieve_top5(query)
prompt = f"""
Question:

{query}

Candidate Documents:

{candidate_docs}

Rank the documents from most relevant
to least relevant. Also provide relevance_score

Return JSON:

ranking : [
  {{
    "document":"...",
    "rank":1,
    "relevance_score":...,
    "reason":"..."
  }}
]
"""
````

````python
response=callGPT(prompt)
````

````python
cleanRankings=return_json(cleanse_response(response),"ranking")
#cleanRanking="\n\n".join(cleanRankings)


print(cleanRankings)
````
