```python
from sentence_transformers import SentenceTransformer

model = SentenceTransformer(
    "all-MiniLM-L6-v2"
)
```

Create simple investigations

```python
investigations = [
    "Inactive Approvers",
    "Workflow Bottleneck",
    "Provisioning Delays",
    "Dormant Accounts",
    "Orphaned Accounts",
    "Certification SLA Failure"
]
```

Generate embeddings

```python
embeddings = model.encode(
    investigations
)
```

```python
print(embeddings.shape)
```

Create Query

```python
query = "Provisioning requests are slow"
```

Generate query embeddings

```python
query_embedding = model.encode(
    query
)
```

Now compute similarity.

```python
from sentence_transformers import util

scores = util.cos_sim(
    query_embedding,
    embeddings
)

print(scores)
```

Find best match

```python
best_index = scores.argmax()

print(investigations[best_index])
```

Try multiple queries

```python
queries = [
    "Managers aren't responding",
    "Access reviews are overdue",
    "Accounts left after termination",
    "Provisioning requests are slow"

]
```

```python
for query in queries:
  query_embedding = model.encode(query)
  scores = util.cos_sim(query_embedding,embeddings)
  print(scores)
  best_index = scores.argmax()
  print(f"{query} : {investigations[best_index]}")
```
