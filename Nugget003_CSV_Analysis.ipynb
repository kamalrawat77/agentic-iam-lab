```python
users = [
    {"name":"Alice","department":"IT","enabled":True},
    {"name":"Bob","department":"IT","enabled":False},
    {"name":"Carol","department":"HR","enabled":True},
    {"name":"David","department":"Finance","enabled":True}
]

print(f"Total Users: {len(users)}")
```

```python
enabled_count = sum(1 for u in users if u["enabled"])

print(f"Enabled Users: {enabled_count}")
```

```python
departments = {}

for user in users:
  dept = user["department"]
  departments[dept] = departments.get(dept,0)+1

print(departments)
```
