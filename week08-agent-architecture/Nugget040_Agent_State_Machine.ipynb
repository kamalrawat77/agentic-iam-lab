Nugget 040: Build Your First Agent State Machine

```python
STATE_PLAN = "PLAN"
STATE_RETRIEVE = "RETRIEVE"
STATE_ANALYZE = "ANALYZE"
STATE_DECIDE = "DECIDE"
STATE_FINISH = "FINISH"
```

```python
current_state = STATE_PLAN
```

```python
while current_state != STATE_FINISH:

    print(
        f"Current State: {current_state}"
    )

    break
```

```python
while current_state != STATE_FINISH:

    if current_state == STATE_PLAN:

        print("Planning")

        current_state = STATE_RETRIEVE


    elif current_state == STATE_RETRIEVE:

        print("Gathering Evidence")

        current_state = STATE_ANALYZE


    elif current_state == STATE_ANALYZE:

        print("Analyzing Findings")

        current_state = STATE_DECIDE


    elif current_state == STATE_DECIDE:

        print("Deciding")

        current_state = STATE_FINISH
```
