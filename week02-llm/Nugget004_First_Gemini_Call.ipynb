```python
from google import genai

client = genai.Client(api_key="API_KEY")

response = client.models.generate_content(
    model="gemini-2.5-flash",
    contents="Explain what Active Directory is in 3 sentences."
)

print(response.text)
```

```python
prompt="Explain SailPoint like I am a new IAM analyst."

response = client.models.generate_content(
    model="gemini-2.5-flash",
    contents=prompt
)

print(response.text)
```
