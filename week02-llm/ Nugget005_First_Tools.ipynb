```python
def count_users(users):
    return len(users)
```

```python
users = [
    {"name":"Alice"},
    {"name":"Bob"},
    {"name":"Carol"}
]

print(count_users(users))
```

```python
def count_enabled_users(users):
    return sum(
        1 for u in users
        if u["enabled"]
    )
```

```python
users = [
    {"name":"Alice","enabled":True},
    {"name":"Bob","enabled":False},
    {"name":"Carol","enabled":True}
]

print(count_enabled_users(users))
```

```python
def department_breakdown(users):

    departments = {}

    for user in users:
        dept = user["department"]

        departments[dept] = (
            departments.get(dept,0) + 1
        )

    return departments
```

```python
users = [
    {"name":"Alice","department":"IT","enabled":True},
    {"name":"Bob","department":"IT","enabled":False},
    {"name":"Carol","department":"HR","enabled":True},
    {"name":"David","department":"Finance","enabled":True}
]
```

```python
print(department_breakdown(users))
```

```python
def disabled_users(users):
    return [
        u["name"]
        for u in users
        if not u["enabled"]
    ]
```

```python
def generate_report(users):

    return {
        "total_users":
            count_users(users),

        "enabled_users":
            count_enabled_users(users),

        "departments":
            department_breakdown(users),

        "disabled_users":
          disabled_users(users)
    }
```

```python
print(generate_report(users))
```
