````python
from google import genai

def callGPT(prompt):
  client = genai.Client(api_key="APIKEY")
  response = client.models.generate_content(
      model="gemini-2.5-flash",
      contents=prompt
  )

  return response
````

````python
def cleanse_response(response):
  clean_response=clean_response = response.text
  clean_response = clean_response.replace("```json", "")
  clean_response = clean_response.replace("```", "")
  clean_response = clean_response.strip()

  return clean_response
````

````python
def return_json(clean_response,objname):
  responseObj = json.loads(clean_response)
  if not objname:
    return responseObj
  resJsonObj  = responseObj[objname]
  return resJsonObj

````

Create dummy tools

````python
def dormant_accounts():

    return "Dormant Accounts: 27"


def department_breakdown():

    return """
IT: 45
HR: 12
Finance: 18
"""


def search_history():

    return """
Past Investigation:
Inactive Approvers
"""


def trend_analysis():

    return """
Dormant Accounts:
20 → 25 → 27
Increasing Trend
"""
````

Create tool registry

````python
tools = {
    "dormant_accounts": dormant_accounts,
    "department_breakdown": department_breakdown,
    "search_history": search_history,
    "trend_analysis": trend_analysis
}
````

Planner prompt

````python
question = """
Have dormant accounts increased
over time?
"""

prompt = f"""
You are an IAM agent.

Available Tools:

- dormant_accounts
- department_breakdown
- search_history
- trend_analysis

Question:

{question}

Return JSON only:

{{
    "tool":"tool_name"
}}
"""
````

````python
response=callGPT(prompt)
````

````python
toolDecision=return_json(cleanse_response(response),None)
````

Execute tool dynamically

````python
selected_tool = toolDecision["tool"]

result = tools[selected_tool]()
````
