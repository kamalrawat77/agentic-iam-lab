````python
from google import genai

def callGPT(prompt):
  client = genai.Client(api_key="APIKEY")
  response = client.models.generate_content(
      model="gemini-2.5-flash",
      contents=prompt
  )

  return response
````

````python
def cleanse_response(response):
  clean_response=clean_response = response.text
  clean_response = clean_response.replace("```json", "")
  clean_response = clean_response.replace("```", "")
  clean_response = clean_response.strip()

  return clean_response
````

````python
observations = [
    "JML completion time increased from 8 to 36 hours",
    "Approval latency increased from 2 to 10 hours",
    "Ticket volume remained stable",
    "Provisioning execution time remained stable"
]
````

````python
prompt = f"""
You are an IAM investigator.

Observations:

{observations}

Generate:

1. Observations
2. Possible causes

Do not identify a root cause yet.
"""
````

````python
print(callGPT(prompt).text)
````

````python
evidence = [
    "Approval latency increased 400%",
    "Inactive approver count increased from 5 to 40",
    "Provisioning system performance unchanged"
]
````

````python
prompt = f"""
Observations:

{observations}

Evidence:

{evidence}

Determine:

1. Most likely root cause
2. Supporting evidence
3. Confidence level

Only use evidence provided.
"""
````

````python
print(callGPT(prompt).text)
````

````python
critic_prompt = f"""
Evidence:

{evidence}

Conclusion:

Inactive approvers causing approval delays.

Review:

1. Is this conclusion justified?
2. What evidence is missing?
3. What would increase confidence?
"""
````

````python
print(callGPT(critic_prompt).text)
````
