````python
from google import genai

def callGPT(prompt):
  client = genai.Client(api_key="APIKEY")
  response = client.models.generate_content(
      model="gemini-2.5-flash",
      contents=prompt
  )

  return response
````

````python
def cleanse_response(response):
  clean_response=clean_response = response.text
  clean_response = clean_response.replace("```json", "")
  clean_response = clean_response.replace("```", "")
  clean_response = clean_response.strip()

  return clean_response
````

````python
#evidence = {
#    "approval_latency": "HIGH",
#    "inactive_approvers": 40,
#    "provisioning_performance": "NORMAL",
#    "hr_feed_delay": False
#}
evidence = {
    "approval_latency": "HIGH",
    "inactive_approvers": 5,
    "provisioning_performance": "SLOW",
    "hr_feed_delay": True
}
````

````python
hypotheses = [
    "Inactive Approvers",
    "Workflow Bottleneck",
    "HR Feed Delay"
]
````

````python
scores = {
    "Inactive Approvers": 0,
    "Workflow Bottleneck": 0,
    "HR Feed Delay": 0
}
````

````python
if evidence["inactive_approvers"] > 20:
    scores["Inactive Approvers"] += 3

if evidence["approval_latency"] == "HIGH":
    scores["Inactive Approvers"] += 2
    scores["Workflow Bottleneck"] += 1

if evidence["hr_feed_delay"]:
    scores["HR Feed Delay"] += 3
````

````python
print(scores)
````

````python
ranked = sorted(
    scores.items(),
    key=lambda x: x[1],
    reverse=True
)

print(ranked)
````
