🚀 Nugget 063 — Structured Outputs and Plan Validation

```python
from pydantic import ValidationError

from scripts.core.planner_schema import PlannerResponse

valid_response = {
    "investigation": "Find dormant accounts",
    "steps": [
        {
            "tool": "dormant_accounts",
            "purpose": "Identify inactive accounts",
            "description": "Find accounts inactive for 90 days",
            "arguments": {
                "days": 90,
                "include_privileged": False,
                "department": "Finance",
            },
        }
    ],
}

parsed = PlannerResponse.model_validate(valid_response)

print(parsed.model_dump())
```

```python
invalid_response = {
    "investigation": "Find dormant accounts",
    "steps": [
        {
            "purpose": "Identify inactive accounts",
            "description": "Find inactive accounts",
            "arguments": {},
        }
    ],
}

try:
    PlannerResponse.model_validate(invalid_response)
except ValidationError as exc:
    print("Validation correctly rejected the response.")
    print(exc)
```
