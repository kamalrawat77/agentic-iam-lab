```python
from google import genai

def callGPT(prompt):
  client = genai.Client(api_key="APIKEY")
  response = client.models.generate_content(
      model="gemini-2.5-flash",
      contents=prompt
  )

  return response
```

```python
report = {
    "orphaned_accounts": 20,
    "dormant_accounts": 35,
    "enabled_users": 430,
    "disabled_users": 70,
    "jml_avg_hours": 36
}
```

```python
def classify_orphaned(count):

    if count < 10:
        return "NORMAL"

    elif count < 50:
        return "WARNING"

    else:
        return "CRITICAL"
```

```python
print(classify_orphaned(25))
```

```python
def classify_jml(hours):

    if hours < 8:
        return "NORMAL"

    elif hours < 24:
        return "WARNING"

    else:
        return "CRITICAL"
```

```python
risk_summary = {
    "orphaned_accounts":
        classify_orphaned(
            report["orphaned_accounts"]
        ),

    "jml_avg_hours":
        classify_jml(
            report["jml_avg_hours"]
        )
}

print(risk_summary)
```

```python
prompt = f"""
You are a senior IAM analyst.

Risk Assessment:

{risk_summary}

Generate:

1. Executive Summary
2. Top Risk
3. Recommended Action

Maximum 5 bullet points.
"""
```

```python
print(callGPT(prompt).text)
```
