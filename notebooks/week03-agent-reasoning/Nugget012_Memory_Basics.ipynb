```python
from google import genai

def callGPT(prompt):
  client = genai.Client(api_key="APIKEY")
  response = client.models.generate_content(
      model="gemini-2.5-flash",
      contents=prompt
  )

  return response
```

```python
question = """
Summarize what we discussed earlier.
"""
```

```python
print(callGPT(question).text)
```

```python
memory = """
Question:
How many enabled users exist?

Answer:
2

Question:
Show department breakdown.

Answer:
IT=2
HR=1
"""
```

```python
prompt = f"""
Conversation History:

{memory}

Current Question:

Summarize what we discussed earlier.
"""
```

```python
print(callGPT(prompt).text)
```

```python
memory = []
question = "How many enabled users are there?"
memory.append({
    "type": "question",
    "value": question
})
result = 2
memory.append({
    "type": "result",
    "value": result
})

memory.append({
    "type": "question",
    "value": "Show department breakdown"
})

memory.append({
    "type": "result",
    "value": "{'IT': 2, 'HR': 1, 'Finance': 1}"
})

def show_memory():
    for item in memory:
        print(item)

show_memory()
```

```python
prompt = f"""
Conversation History:

{memory}

Current Question:

Summarize what we discussed earlier.
"""
```

```python
print(callGPT(prompt).text)
```
