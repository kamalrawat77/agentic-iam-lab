````python
def count_users(users):
    return len(users)
def count_enabled_users(users):
    return sum(
        1 for u in users
        if u["enabled"]
    )
def department_breakdown(users):

    departments = {}

    for user in users:
        dept = user["department"]

        departments[dept] = (
            departments.get(dept,0) + 1
        )

    return departments
def disabled_users(users):
    return [
        u["name"]
        for u in users
        if not u["enabled"]
    ]
def generate_report(users):

    return {
        "total_users":
            count_users(users),

        "enabled_users":
            count_enabled_users(users),

        "departments":
            department_breakdown(users),

        "disabled_users":
          disabled_users(users)
    }
````

````python
from google import genai

def callGPT(prompt):
  client = genai.Client(api_key="API")
  response = client.models.generate_content(
      model="gemini-2.5-flash",
      contents=prompt
  )

  return response
````

````python
available_tools = [
    "count_users",
    "count_enabled_users",
    "department_breakdown",
    "disabled_users"
]
````

````python
#question = "How many enabled users are there?"
#question = "Give me the name of disabled users"
#question = "Can you please provide me the details about department"
#question = "Which business units have users?"
question = """
Provide an IAM health summary.
"""
````

````python
planning_prompt = f"""
You are an IAM analyst.

Available tools:

{available_tools}

Question:

{question}

Return JSON only.

Example:

{{
  "tools": [
    "count_users",
    "count_enabled_users"
  ]
}}
"""

llmResponse = callGPT(planning_prompt)
````
