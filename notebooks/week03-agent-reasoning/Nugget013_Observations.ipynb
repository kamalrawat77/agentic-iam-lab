```python
from google import genai

def callGPT(prompt):
  client = genai.Client(api_key="APIKEY")
  response = client.models.generate_content(
      model="gemini-2.5-flash",
      contents=prompt
  )

  return response
```

```python
report = {
    "orphaned_accounts": 20,
    "dormant_accounts": 35,
    "enabled_users": 430,
    "disabled_users": 70
}
```

```python
prompt = f"""
You are an IAM analyst.

Report Data:

{report}

Generate:

1. Key Observations
2. Risks
3. Recommended Actions

Limit to 5 bullet points.
"""
```

```python
print(callGPT(prompt).text)
```

```python
prompt = f"""
You are a senior IAM analyst.

Report Data:

{report}

Rules:

- Focus on identity risk
- Mention trends if visible
- Mention operational concerns
- Provide actionable recommendations

Output bullets only.
"""
```

```python
print(callGPT(prompt).text)
```
