```python
from scripts.agent.planner import Planner
from scripts.tools.registry import AVAILABLE_TOOLS

planner = Planner()

question = """
Why are dormant accounts increasing?
"""

plan = planner.create_plan(
    question,
    AVAILABLE_TOOLS
)

print(plan)
```

Validate the Plan

```python
tool_names = {
    tool["name"]
    for tool in AVAILABLE_TOOLS
}

for step in plan["steps"]:

    if step["tool"] not in tool_names:

        raise ValueError(
            f"Unknown tool: {step['tool']}"
        )
```
