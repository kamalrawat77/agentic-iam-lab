Nugget 043: Build the Investigation Planner

````python
!pip install -q google-genai
import json
````

````python
from google import genai

def callGPT(prompt):
  client = genai.Client(api_key="APIKEY")
  response = client.models.generate_content(
      model="gemini-2.5-flash",
      contents=prompt
  )

  return response
````

````python
def cleanse_response(response):
  clean_response=clean_response = response.text
  clean_response = clean_response.replace("```json", "")
  clean_response = clean_response.replace("```", "")
  clean_response = clean_response.strip()

  return clean_response
````

````python
def return_json(clean_response,objname):
  responseObj = json.loads(clean_response)
  if not objname:
    return responseObj
  resJsonObj  = responseObj[objname]
  return resJsonObj

````

Define Available Tools

````python
available_tools = [
    {
        "name": "trend_analysis",
        "description": "Analyze trends over time"
    },
    {
        "name": "search_history",
        "description": "Search previous investigations"
    },
    {
        "name": "department_breakdown",
        "description": "Summarize department statistics"
    },
    {
        "name": "dormant_accounts",
        "description": "Return dormant account metrics"
    }
]

tools = [
    "trend_analysis",
    "search_history",
    "department_breakdown",
    "dormant_accounts"
]
````

Planner Prompt

````python
question = "Why are dormant accounts increasing?"
````

````python
planner_prompt = f"""
You are an IAM Investigation Planner.

Question:

{question}

Available Tools:

{available_tools}

Your job is NOT to answer the question.

Your job is to create an investigation plan.

For each step explain:

- tool
- purpose

Return ONLY JSON.

Example:

{{
  "investigation":"...",
  "steps":[
    {{
      "tool":"trend_analysis",
      "purpose":"...",
      "priority":1
    }}
  ]
}}
"""
````

````python
response=callGPT(planner_prompt)
plan=return_json(cleanse_response(response),None)
````

Parse the Plan

````python
print(plan["investigation"])

for step in plan["steps"]:
    print(step["tool"])
    print(step["purpose"])
    print(step["priority"])
````

Validate the Plan

````python
for step in plan["steps"]:

    if step["tool"] not in tools:

        print(f"Unknown tool: {step['tool']}")
````
