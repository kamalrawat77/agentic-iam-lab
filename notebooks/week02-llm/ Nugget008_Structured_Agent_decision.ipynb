````python
def count_users(users):
    return len(users)
def count_enabled_users(users):
    return sum(
        1 for u in users
        if u["enabled"]
    )
def department_breakdown(users):

    departments = {}

    for user in users:
        dept = user["department"]

        departments[dept] = (
            departments.get(dept,0) + 1
        )

    return departments
def disabled_users(users):
    return [
        u["name"]
        for u in users
        if not u["enabled"]
    ]
def generate_report(users):

    return {
        "total_users":
            count_users(users),

        "enabled_users":
            count_enabled_users(users),

        "departments":
            department_breakdown(users),

        "disabled_users":
          disabled_users(users)
    }
````

````python
from google import genai

def callGPT(prompt):
  client = genai.Client(api_key="APIKEY")
  response = client.models.generate_content(
      model="gemini-2.5-flash",
      contents=prompt
  )

  return response
````

````python
#question = "How many enabled users are there?"
#question = "Give me the name of disabled users"
#question = "Can you please provide me the details about department"
question = "Which business units have users?"
````

````python
prompt = f"""
You are an IAM assistant.

Available tools:

- count_users
- count_enabled_users
- department_breakdown
- disabled_users

Question:

{question}

Return JSON only.

Example:

{{"tool":"count_users"}}
"""

llmResponse = callGPT(prompt)
````

````python
users = [
    {"name":"Alice","department":"IT","enabled":True},
    {"name":"Bob","department":"IT","enabled":False},
    {"name":"Carol","department":"HR","enabled":True},
    {"name":"David","department":"Finance","enabled":True}
]
````

````python
print(llmResponse.text)
````

````python
def cleanse_response(response):
  clean_response=clean_response = response.text
  clean_response = clean_response.replace("```json", "")
  clean_response = clean_response.replace("```", "")
  clean_response = clean_response.strip()

  return clean_response
````

````python
import json



selection = json.loads(cleanse_response(llmResponse))

selected_tool = selection["tool"]

print(selected_tool)
````

````python
if selected_tool == "count_users":
    result = count_users(users)

elif selected_tool == "count_enabled_users":
    result = count_enabled_users(users)

elif selected_tool == "department_breakdown":
    result = department_breakdown(users)

elif selected_tool == "disabled_users":
    result = disabled_users(users)

print(result)
````
