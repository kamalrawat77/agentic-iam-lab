```python
def count_users(users):
    return len(users)
def count_enabled_users(users):
    return sum(
        1 for u in users
        if u["enabled"]
    )
def department_breakdown(users):

    departments = {}

    for user in users:
        dept = user["department"]

        departments[dept] = (
            departments.get(dept,0) + 1
        )

    return departments
def disabled_users(users):
    return [
        u["name"]
        for u in users
        if not u["enabled"]
    ]
def generate_report(users):

    return {
        "total_users":
            count_users(users),

        "enabled_users":
            count_enabled_users(users),

        "departments":
            department_breakdown(users),

        "disabled_users":
          disabled_users(users)
    }
```

```python
from google import genai

def callGPT(prompt):
  client = genai.Client(api_key="API_KEY")
  response = client.models.generate_content(
      model="gemini-2.5-flash",
      contents=prompt
  )

  return response.text.strip()
```

```python
#question = "How many enabled users are there?"
#question = "Give me the name of disabled users"
question = "Can you please provide me the details about department"
```

```python
prompt = f"""
You are an IAM assistant.

Available tools:

1. count_users
2. count_enabled_users
3. department_breakdown
4. disabled_users

Question:

{question}

Return ONLY the tool name.
"""

selected_tool = callGPT(prompt)
```

```python
users = [
    {"name":"Alice","department":"IT","enabled":True},
    {"name":"Bob","department":"IT","enabled":False},
    {"name":"Carol","department":"HR","enabled":True},
    {"name":"David","department":"Finance","enabled":True}
]
```

```python
if selected_tool == "count_users":
    result = count_users(users)

elif selected_tool == "count_enabled_users":
    result = count_enabled_users(users)

elif selected_tool == "department_breakdown":
    result = department_breakdown(users)

elif selected_tool == "disabled_users":
    result = disabled_users(users)

print(result)
```
