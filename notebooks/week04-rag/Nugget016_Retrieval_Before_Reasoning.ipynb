```python
from google import genai

def callGPT(prompt):
  client = genai.Client(api_key="APIKEY")
  response = client.models.generate_content(
      model="gemini-2.5-flash",
      contents=prompt
  )

  return response
```

```python
knowledge_base = [
    "Workday aggregation failed on Tuesday.",
    "Orphaned accounts increased from 10 to 20.",
    "JML completion time increased from 6 to 24 hours.",
    "Source health returned to GREEN on Wednesday."
]
```

```python
def retrieve_context(question):

    results = []

    for item in knowledge_base:

        if "orphaned" in question.lower() and "orphaned" in item.lower():
            results.append(item)

    return results
```

```python
retrieve_context(
    "Why are orphaned accounts increasing?"
)
```

```python
question = "Why are orphaned accounts increasing?"
context = retrieve_context(question)

prompt = f"""
Evidence:

{context}

Question:

{question}

Use only the evidence provided.
"""
```

```python
print(callGPT(prompt).text)
```
