```python
from google import genai

def callGPT(prompt):
  client = genai.Client(api_key="APIKEY")
  response = client.models.generate_content(
      model="gemini-2.5-flash",
      contents=prompt
  )

  return response
```

```python
knowledge_base = [
    "Orphaned accounts increased from 10 to 20.",
    "Workday aggregation failed Tuesday.",
    "HR termination feed was delayed.",
    "JML processing increased from 4 to 18 hours."
]
```

```python
def retrieve(keyword):

    return [
        item
        for item in knowledge_base
        if keyword.lower() in item.lower()
    ]
```

```python
question = """
Why are orphaned accounts increasing?
"""

prompt = f"""
Question:

{question}

Before answering:

What additional information would help?
"""
```

```python
print(callGPT(prompt).text)
```

```python
additional_evidence = retrieve("Workday")
```

```python
prompt = f"""
Question:

{question}

Evidence:

{additional_evidence}

Provide an answer.
"""
```

```python
print(callGPT(prompt).text)
```
